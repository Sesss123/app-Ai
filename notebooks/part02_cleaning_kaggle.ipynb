# TripMe Part 2 — Canonical Data Cleaning
Attach the `tripme-part02-cleaning` dataset and run all cells. Raw inputs remain unchanged; no GPU is required.

```python
from pathlib import Path
import shutil

scripts = list(Path('/kaggle/input').rglob('scripts/part02_clean_places.py'))
if not scripts:
    raise FileNotFoundError('Attach the tripme-part02-cleaning Kaggle Dataset')
input_root = scripts[0].parent.parent
raw_dir = input_root / 'data' / 'raw'
output_dir = Path('/kaggle/working/part02_cleaning_output')
if output_dir.exists():
    shutil.rmtree(output_dir)
output_dir.mkdir(parents=True)
print('Input:', input_root)
print('Raw files:', len(list(raw_dir.rglob('*.jsonl'))))
```

```python
import subprocess
import sys

subprocess.run([
    sys.executable, str(scripts[0]),
    '--raw-dir', str(raw_dir),
    '--output-dir', str(output_dir),
], check=True)
```

```python
import json

summary = json.loads((output_dir / 'cleaning_summary.json').read_text(encoding='utf-8'))
assert summary['raw_files'] == 363, summary
assert summary['records'] > 0, summary
assert summary['parse_errors'] == 0, summary
print(json.dumps(summary, indent=2))
```
